### 1. Executive Summary & Methodology
In Stage 3 of our data mining lifecycle, we perform in-depth Exploratory Data Analysis (EDA) and apply three distinct feature selection techniques:
1. **Univariate Linear Correlation Filter (Pearson)**: Identifies monotonic linear associations with target `is_adulterated`.
2. **Chi-Square ($\chi^2$) Test of Independence**: Measures statistical independence between non-negative transformed attributes and binary target classes.
3. **Random Forest Gini Importance (MDI)**: Evaluates non-linear feature splits and interaction importance across an ensemble of decision trees.
