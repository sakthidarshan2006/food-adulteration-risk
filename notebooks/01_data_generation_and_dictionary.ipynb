5. **Honey** (Sugar syrup, high-fructose corn syrup, moisture limit breaches)
6. **Beverages** (Non-permitted artificial sweeteners, toxic synthetic colorants)
7. **Processed Foods** (Recycled frying fats, chemical leavening agents)

### 3. Sensory Rubric & Critical Assessment of Subjectivity
In the field, food safety officers frequently perform immediate sensory organoleptic assessments before selecting samples for forensic laboratory dispatch:

| Score | Description | Technical Rubric |
|---|---|---|
| **1 (Severe Defect)** | Acute deterioration / Chemical adulterant | Obvious presence of foreign dyes, pungent chemical or sour odor, off-taste, punctured/tampered seal |
| **2 (Moderate Defect)** | Noticeable deviation | Dull/unnatural tint, stale odor, flat/watery taste, worn or creased packaging |
| **3 (Borderline)** | Ambiguous conformity | Color slightly off-season, faint odor loss, slightly abnormal mouthfeel, minor packaging scuffs |
| **4 (Acceptable)** | Standard commercial grade | Conforming appearance, expected aroma, standard taste, intact packaging |
| **5 (Prime)** | Superior quality | Ideal characteristic color, fragrant fresh aroma, pristine taste, hermetically sealed |

> 🔍 **Academic Critical Note on Sensory Limitation:**  
> Sensory scores inherently introduce **inter-rater variability** and subjective human bias. An inspector fatigued at the end of a shift may rate samples differently from one in the morning. In Phase 10 (Future Enhancements), we discuss how hyperspectral imaging and electronic-nose (e-nose) IoT sensors can replace subjective sensory scoring with objective digitized feature vectors.
